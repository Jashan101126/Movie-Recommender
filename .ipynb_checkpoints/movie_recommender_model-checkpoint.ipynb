1. Project Overview

This project is a Content-Based Movie Recommender System that suggests movies similar to a given movie based on their features.

The system uses metadata such as genres, keywords, cast, crew, and movie overview to find similarities between movies. It applies Natural Language Processing (NLP) techniques and cosine similarity to generate recommendations.

Technologies Used:

Python
Pandas, NumPy
Scikit-learn
NLP
Pickle

2. Dataset Information

The dataset used in this project is movies_metadata.csv, which contains detailed metadata for thousands of movies.

Key Features Used:

title – Name of the movie
overview – Description/summary of the movie
genres – List of genres
keywords – Tags associated with the movie (if used)
vote_average – Average rating
popularity – Popularity score

3. Importing Libraries

All necessary libraries are imported for data handling, preprocessing, and building the recommendation system.

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
import ast

import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import difflib

import warnings
warnings.filterwarnings("ignore")
```

4. Loading the Data

The dataset is loaded using Pandas as the dataset is large and may contain mixed data types. 

```python
df = pd.read_csv("movies_metadata.csv")
df.head()
```

5. Data Understanding (EDA) 

```python
print(df.shape)
print(df.info())
print(df.isnull().sum())
df.info()
```

6. Data Cleaning and Handling missing data 

```python
df.dropna(inplace=True)
df = df.drop_duplicates().reset_index(drop= True)
df.isnull().sum() 
df['overview'] = df['overview'].fillna("")
df['tagline'] = df['tagline'].fillna("")
df = df.dropna(subset= ['title'])
df.head()
```

7. Feature Engineering

```python
df['genres'] = df['genres'].apply (lambda x : " ".join([i['name']for i in ast.literal_eval(x)]))
```

8. Creating Tags 

```python
df['tags'] = df['title'] + " " + df['overview'] + " " + df['genres'] + " "+ df['tagline']
```

9. Text Preprocessing

```python
nltk.download("stopwords")
nltk.download("wordnet")
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
def preprocess_text(text):
    text = str(text).lower()
    text = re.sub(r'[^\w\s]', '', text) # removing punctuations
    words = text.split()
    words = [word for word in words if word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words] 

    return " ".join(words)

df['tags'] = df['tags'].apply(preprocess_text)
df['tags'][1]
```

```python
df.reset_index(inplace=True)
```

10. Vectorization

```python
indices = pd.Series(df.index, index=df['title'].str.lower()).drop_duplicates()
tfidf = TfidfVectorizer(max_features= 50000 , ngram_range=(1,2), stop_words= "english")
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

11. Cosine Similarity 

```python
tfidf_matrix.shape
```

12. Recommendation Function

```python
def recommend(title , n = 10): 
    title = title.lower().strip()
    if title not in indices:
        matches = difflib.get_close_matches(title ,indices.index,n=1,cutoff=0.6)
        if matches:
            return [f"Movie not found. Did you mean '{matches[0]}'?"]
        return ["Movie not Found"]

    idx = indices[title]
    sim_score = cosine_similarity(tfidf_matrix[idx:idx+1], tfidf_matrix).flatten()
    similar_idx = sim_score.argsort()[::-1][1:n+1]

    recommendations = []
    for i in similar_idx:
        if sim_score[i] < 0.15:
            continue
        recommendations.append((df['title'].iloc[i], round(float(sim_score[i]), 3)))
    
    return recommendations
```

13. Test

```python
recommend('jumanji')
```
